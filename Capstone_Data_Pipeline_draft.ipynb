```python
import pandas as pd
from datetime import timedelta
```

# Modify the residential datasets

#Single Family Detached

```python
input_file_path ='C:/00 Grand Canyon University/Capstone/Datasets/Residential/15_minute_timeseries_data_single_fam_detached.csv'  
output_file_path ='C:/00 Grand Canyon University/Capstone/Datasets/Database/modified_15_minute_timeseries_data_single_fam_detached.csv'   
```

#Read the CSV file

```python
df = pd.read_csv(input_file_path)
```

#Select only the desired columns

```python
df = df[['Timestamp (EST)', 'baseline.out.site_energy.total.energy_consumption.kwh']]
```

#Rename the columns

```python
df.rename(columns={
    'Timestamp (EST)': 'timestamp_EST',
    'baseline.out.site_energy.total.energy_consumption.kwh': 'total_energy_consumption_single_fam_detached_kwh'
}, inplace=True)
```

```python
print(df.dtypes)
```

```python
print(df.shape)
```

```python
df['timestamp_EST'] = pd.to_datetime(df['timestamp_EST'])
```

```python
print(df.dtypes)
```

#Set the timestamp as the index

```python
df.set_index('timestamp_EST', inplace=True)
```

#Resample the data to calculate hourly average consumption

```python
hourly_avg = df['total_energy_consumption_single_fam_detached_kwh'].resample('h').mean()
```

#Map the hourly average back to the original DataFrame

```python
df['hourly_consumption_single_fam_detached_kwh'] = df.index.map(hourly_avg)
```

#Reset index

```python
df.reset_index(inplace=True)
```

#Drop rows with NaN values

```python
df.dropna(inplace=True)
```

#Drop the original total energy consumption column

```python
df.drop(columns=['total_energy_consumption_single_fam_detached_kwh'], inplace=True)
```

#Reset the index to start from 0

```python
df.reset_index(drop=True, inplace=True)
```

```python
print(df.head())
```

```python
print(df.shape)
```

#Save the modified DataFrame to a new CSV file

```python
df.to_csv(output_file_path, index=False)
```

# Modify Weather Data

```python
import pandas as pd
```

```python
input_file_path ='C:/00 Grand Canyon University/Capstone/Datasets/Weather Dataset/texas_weather.csv'
output_file_path = 'C:/00 Grand Canyon University/Capstone/Datasets/Database/modified_texas_weather.csv'
```

#Read the CSV file

```python
df = pd.read_csv(input_file_path)
```

#Keep only the desired columns

```python
df = df[['dt_iso', 'temp', 'humidity']]
```

```python
print(df.head())
```

```python
print(df.dtypes)
```

#Remove the timezone string

```python
df['dt_iso'] = df['dt_iso'].str.replace('+0000 UTC', '', regex=False)
```

#Convert to datetime and localize to UTC

```python
df['dt_iso'] = pd.to_datetime(df['dt_iso']).dt.tz_localize('UTC')
```

#Convert to America/New_York timezone

```python
df['dt_iso'] = df['dt_iso'].dt.tz_convert('America/New_York')
```

#Convert to UTC-5

```python
df['dt_iso'] = df['dt_iso'].dt.tz_convert('UTC-05:00')
```

```python
print(df.head())
```

#Set the start date

```python
start_date = pd.Timestamp('2018-01-01 01:00:00-05:00')
```

#Filter the DataFrame to include only dates from the start date onward

```python
df = df[df['dt_iso'] >= start_date]
```

#Reset the index to start from 0

```python
df.reset_index(drop=True, inplace=True)
```

```python
print(df.head())
```

```python
print(df.dtypes)
```

```python
print(df.shape)
```

```python
df.rename(columns={
    'dt_iso': 'timestamp_EST',
    'temp': 'temp_TX',  
    'humidity': 'humidity_TX',
}, inplace=True)
```

```python
print(df.head())
```

```python
print(df.dtypes)
```

#Save the cleaned DataFrame

```python
df.to_csv(output_file_path, index=False)
```

# Data Ingestion

```python
import pandas as pd
import os
```

#Specify the directory containing the CSV files and initialize an empty DataFrame

```python
directory = 'C:/00 Grand Canyon University/Capstone/Datasets/Database'
merged_df = pd.DataFrame() 
```

#Loop through all files in the specified directory

```python
for filename in os.listdir(directory):
    if filename.endswith('.csv'):
        file_path = os.path.join(directory, filename)
        
        # Read the CSV file, specifying low_memory=False to avoid DtypeWarning
        df = pd.read_csv(file_path, low_memory=False)
        
        # Merge the current DataFrame with the merged DataFrame
        merged_df = pd.concat([merged_df, df], axis=1)

# Drop duplicate columns, keeping the first occurrence
merged_df = merged_df.loc[:, ~merged_df.columns.duplicated()]
```

# Visualization

#Exploring/Visualizing our data

```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

#Read the CSV file

```python
file_path = 'C:/00 Grand Canyon University/Capstone/Datasets/MergedDatabase/merged_output.csv'
df = pd.read_csv(file_path)
```

#Define the explore_data function

```python
def explore_data(df):
    # Basic statistics
    print(df.describe())
    
    # Select numerical columns excluding datetime columns
    numerical_columns = df.select_dtypes(include=['float64']).columns
    
    # Visualize distributions for all numerical columns
    for column in numerical_columns:
        plt.figure(figsize=(12, 6))
        sns.histplot(df[column], bins=30, kde=True)
        plt.title(f'Distribution of {column}')
        plt.xlabel(column)
        plt.ylabel('Frequency')
        plt.show()
```

```python
def plot_correlation_heatmap(df):
    # Select only numerical columns
    numerical_df = df.select_dtypes(include=['number'])

    # Create the correlation heatmap
    plt.figure(figsize=(10, 8))
    sns.heatmap(numerical_df.corr(), annot=True, fmt=".2f", cmap='coolwarm')
    plt.title('Correlation Heatmap')
    plt.show()
```

#Call the explore_data function

```python
explore_data(df)
```

#Call the plot_correlation_heatmap function

```python
plot_correlation_heatmap(df)
```
